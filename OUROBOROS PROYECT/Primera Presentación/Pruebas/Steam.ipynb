```python
reviews_map = df_reviews.set_index('AppID')['Total_Reviews']
reviews_zero = (df['Total_Reviews'] == 0)
appids_to_update = pd.Index(df.loc[reviews_zero, 'AppID'].dropna().astype(int)).intersection(reviews_map.index)
df.loc[df['AppID'].isin(appids_to_update) & reviews_zero, 'Total_Reviews'] = \
    df.loc[df['AppID'].isin(appids_to_update) & reviews_zero, 'AppID'].map(reviews_map)
```

```python
df.loc[df['Estimated_Owners'] == 0, 'Estimated_Owners'] = df['Total_Reviews'] * 30
```

```python
df['Total_Sales'] = df['Estimated_Owners']
df['Revenue'] = df['Total_Sales']*df['Price_USD']
```

```python
condition1 = df['Total_Reviews'] != 0
Juegos_reviews_actualizado = df[condition1]
display(Juegos_reviews_actualizado.head(1000))
```

```python
juegos_gratis = Juegos_reviews_actualizado[Juegos_reviews_actualizado['Price_USD'] == 0]
juegos_gratis.head(107)
```

```python
ARPU_base = {"Action": 13, "Adventure": 9, "RPG": 20, "Strategy": 13, "Simulation": 8, "Casual": 6, "Massively Multiplayer": 22, "Indie": 7, "Sports": 11, "Racing": 9, "Unknown": 10}
```

```python
Juegos_reviews_actualizado['ARPU'] = Juegos_reviews_actualizado['Primary_Genre'].map(ARPU_base)
free_games = (Juegos_reviews_actualizado['Price_USD'] == 0)
Juegos_reviews_actualizado.loc[free_games,'Revenue'] = Juegos_reviews_actualizado.loc[free_games,'Estimated_Owners'] * Juegos_reviews_actualizado.loc[free_games,'ARPU']

Juegos_reviews_actualizado.head(1000)
```

```python
import plotly.express as px
```

```python
ventas_por_genero = Juegos_reviews_actualizado.groupby('Primary_Genre')['Revenue'].sum().reset_index()

fig_bar = px.bar(
    ventas_por_genero,
    x='Primary_Genre',
    y='Revenue',
    title='Ventas Totales por Género Principal',
    labels={'Primary_Genre': 'Género Principal', 'Revenue': 'Ventas Totales'},
    hover_data={'Revenue': ':.2s'}
)

fig_bar.update_layout(xaxis_title='Género Principal', yaxis_title='Ventas Totales')
fig_bar.show()
```

```python
fig_pie = px.pie(
    ventas_por_genero,
    values='Revenue',
    names='Primary_Genre',
    title='Porcentaje de Ingresos por Género Principal',
    labels={'Primary_Genre': 'Género Principal', 'Revenue': 'Ingresos'},
    hole=0.0 # Optional: creates a donut chart
)

fig_pie.update_traces(textposition='inside', textinfo='percent+label')
fig_pie.show()
```

```python
ventas_totales_por_genero = Juegos_reviews_actualizado.groupby('Primary_Genre')['Total_Sales'].sum().reset_index()

fig_bar_sales = px.bar(
    ventas_totales_por_genero,
    x='Primary_Genre',
    y='Total_Sales',
    title='Ventas Totales de Juegos por Género Principal',
    labels={'Primary_Genre': 'Género Principal', 'Total_Sales': 'Ventas Totales'},
    hover_data={'Total_Sales': ':.2s'}
)

fig_bar_sales.update_layout(xaxis_title='Género Principal', yaxis_title='Ventas Totales')
fig_bar_sales.show()
```

```python
fig_pie_sales = px.pie(
    ventas_totales_por_genero,
    values='Total_Sales',
    names='Primary_Genre',
    title='Porcentaje de Ventas por Género Principal',
    labels={'Primary_Genre': 'Género Principal', 'Total_Sales': 'Ventas Totales'},
    hole=0.0 # Optional: creates a donut chart
)

fig_pie_sales.update_traces(textposition='inside', textinfo='percent+label')
fig_pie_sales.show()
```
