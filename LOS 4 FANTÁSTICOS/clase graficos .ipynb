```python
# DataFrame del período
df_berlin = df[(df['year'] >= 1984) & (df['year'] <= 1994)].copy()

# Eliminamos países sin clasificación regional
todos_clasificados = (america_del_sur + america_central_y_caribe + america_del_norte +
                      europa + africa + asia + oceania)
df_berlin = df_berlin[df_berlin['country'].isin(todos_clasificados)]

# Asignación de región geográfica
def asignar_region(pais):
    if pais in america_del_sur:          return 'América del Sur'
    if pais in america_central_y_caribe: return 'América Central y Caribe'
    if pais in america_del_norte:        return 'América del Norte'
    if pais in europa:                   return 'Europa'
    if pais in africa:                   return 'África'
    if pais in asia:                     return 'Asia'
    if pais in oceania:                  return 'Oceanía'


# Asignación de bloque económico
def asignar_bloque(pais):
    if pais in ocde:          return 'OCDE'
    if pais in union_europea: return 'Unión Europea'
    if pais in brics:         return 'BRICS'
    if pais in mercosur:      return 'MERCOSUR'
    if pais in medio_oriente: return 'Medio Oriente'
    return 'Otros'

df_berlin['region'] = df_berlin['country'].apply(asignar_region)
df_berlin['bloque'] = df_berlin['country'].apply(asignar_bloque)

# Sub-período
def etiqueta_periodo(year):
    if year <= 1988: return 'Antes (1984-1988)'
    if year <= 1991: return 'Evento (1989-1991)'
    return 'Después (1992-1994)'

df_berlin['sub_periodo'] = df_berlin['year'].apply(etiqueta_periodo)
orden_periodos = ['Antes (1984-1988)', 'Evento (1989-1991)', 'Después (1992-1994)']

# Sub-DataFrames por región
df_muro_sur      = df_berlin[df_berlin['country'].isin(america_del_sur)]
df_muro_central  = df_berlin[df_berlin['country'].isin(america_central_y_caribe)]
df_muro_norte    = df_berlin[df_berlin['country'].isin(america_del_norte)]
df_muro_europa   = df_berlin[df_berlin['country'].isin(europa)]
df_muro_africa   = df_berlin[df_berlin['country'].isin(africa)]
df_muro_asia     = df_berlin[df_berlin['country'].isin(asia)]
df_muro_oceania  = df_berlin[df_berlin['country'].isin(oceania)]

# Sub-DataFrames por bloque
df_muro_mercosur      = df_berlin[df_berlin['country'].isin(mercosur)]
df_muro_brics         = df_berlin[df_berlin['country'].isin(brics)]
df_muro_medio_oriente = df_berlin[df_berlin['country'].isin(medio_oriente)]
df_muro_ocde          = df_berlin[df_berlin['country'].isin(ocde)]
df_muro_ue            = df_berlin[df_berlin['country'].isin(union_europea)]

print(f'✅ DataFrame período: {df_berlin.shape[0]} registros | {df_berlin["country"].nunique()} países')
df_berlin.head()
```

```python
# Tabla por región y sub-período
tabla_region = df_berlin.groupby(['region', 'sub_periodo']).agg(
    gdp_per_capita_prom  = ('gdp_per_capita',  'mean'),
    income_top10_prom    = ('income_top10',    'mean'),
    income_bottom50_prom = ('income_bottom50', 'mean'),
    income_top1_prom     = ('income_top1',     'mean'),
).round(2).reset_index()
tabla_region['sub_periodo'] = pd.Categorical(
    tabla_region['sub_periodo'], categories=orden_periodos, ordered=True)
tabla_region = tabla_region.sort_values(['region', 'sub_periodo'])

print('📊 Estadísticas por Región y Sub-período')
display(tabla_region)
```

```python
pivot = tabla_region.pivot_table(
    index='region', columns='sub_periodo', values='gdp_per_capita_prom')
pivot.columns.name = None

resumen = pd.DataFrame()
resumen['PBI pc — Antes']   = pivot['Antes (1984-1988)'].round(0)
resumen['PBI pc — Después'] = pivot['Después (1992-1994)'].round(0)
resumen['Variación %'] = (
    (pivot['Después (1992-1994)'] - pivot['Antes (1984-1988)'])
    / pivot['Antes (1984-1988)'] * 100
).round(1).astype(str) + '%'

print('📋 Variación del PBI per cápita — Antes vs Después')
display(resumen.sort_values('PBI pc — Después', ascending=False))
```

```python
antes   = df_berlin[df_berlin['sub_periodo'] == 'Antes (1984-1988)'].groupby('region')['gdp_per_capita'].mean()
despues = df_berlin[df_berlin['sub_periodo'] == 'Después (1992-1994)'].groupby('region')['gdp_per_capita'].mean()

comp = pd.DataFrame({'antes': antes, 'despues': despues}).dropna().reset_index()
comp = comp[comp['region'] != 'Sin clasificar']
comp['var_pct'] = ((comp['despues'] - comp['antes']) / comp['antes'] * 100).round(1)
comp = comp.sort_values('var_pct', ascending=True)

import plotly.graph_objects as go

fig4 = go.Figure(go.Bar(
    x=comp['var_pct'], y=comp['region'], orientation='h',
    marker_color=['#d62728' if v < 0 else '#2ca02c' for v in comp['var_pct']],
    text=[f'{v:+.1f}%' for v in comp['var_pct']], textposition='outside'))
fig4.add_vline(x=0, line_color='black', line_width=1)
fig4.update_layout(
    title='📊 Variación % del PBI per cápita — Antes (1984-88) vs Después (1992-94)',
    xaxis_title='Variación %', yaxis_title='Región',
    template='plotly_white', height=400)
fig4.show()
```

Para mejorar la claridad visual, a continuación se presenta un gráfico de barras agrupadas que compara directamente los valores absolutos del PBI per cápita 'Antes' y 'Después' para cada región. Esto permite una comparación más intuitiva de la evolución económica.

```python
fig_grouped_gdp = go.Figure(data=[
    go.Bar(name='PBI pc — Antes (1984-1988)', x=comp['region'], y=comp['antes']),
    go.Bar(name='PBI pc — Después (1992-1994)', x=comp['region'], y=comp['despues'])
])

fig_grouped_gdp.update_layout(barmode='group',
                    title='📈 PBI per cápita promedio por Región — Antes vs Después (Valores Absolutos)',
                    xaxis_title='Región',
                    yaxis_title='PBI per cápita (USD)',
                    template='plotly_white',
                    height=500)
fig_grouped_gdp.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
ingreso = df_berlin[df_berlin['region'] != 'Sin clasificar'].groupby(
    ['region', 'sub_periodo']).agg(
    top10 = ('income_top10',    'mean'),
    bot50 = ('income_bottom50', 'mean')
).reset_index()
ingreso['sub_periodo'] = pd.Categorical(
    ingreso['sub_periodo'], categories=orden_periodos, ordered=True)
ingreso = ingreso.sort_values('sub_periodo')

fig3 = make_subplots(rows=1, cols=2,
    subplot_titles=('% ingreso del Top 10%', '% ingreso del Bottom 50%'))

for region in ingreso['region'].unique():
    d = ingreso[ingreso['region'] == region]
    fig3.add_trace(go.Bar(name=region, x=d['sub_periodo'], y=d['top10'],
        legendgroup=region, showlegend=True), row=1, col=1)
    fig3.add_trace(go.Bar(name=region, x=d['sub_periodo'], y=d['bot50'],
        legendgroup=region, showlegend=False), row=1, col=2)

fig3.update_layout(
    title_text='💰 Distribución del ingreso por región — Antes, Durante y Después',
    barmode='group', template='plotly_white', height=450)
fig3.show()
```
