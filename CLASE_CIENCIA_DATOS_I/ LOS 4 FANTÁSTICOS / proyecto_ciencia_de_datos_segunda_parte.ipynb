```python
import pandas as pd
df = pd.read_csv('disoccupazione - disoccupazione.csv')
```

```python
df_15_plus = df[df['age'] == '15+']
df_15_plus
```

```python
# Filter out unrealistic unemployment rates (e.g., values > 100%) from df_15_plus
df_15_plus_filtered = df_15_plus[df_15_plus['obs_value'] <= 100]

# Top 5 de mayor tasa de desempleo para hombres (filtrado de df_15_plus)
top_5_male_15_plus_filtered = df_15_plus_filtered[df_15_plus_filtered['sex'] == 'Male'].sort_values(by='obs_value', ascending=False).head(5)
print('Top 5 de mayor tasa de desempleo para Hombres (en df_15_plus, excluyendo valores atípicos):')
print(top_5_male_15_plus_filtered)

print('\n')

# Top 5 de mayor tasa de desempleo para mujeres (filtrado de df_15_plus)
top_5_female_15_plus_filtered = df_15_plus_filtered[df_15_plus_filtered['sex'] == 'Female'].sort_values(by='obs_value', ascending=False).head(5)
print('Top 5 de mayor tasa de desempleo para Mujeres (en df_15_plus, excluyendo valores atípicos):')
print(top_5_female_15_plus_filtered)
```

```python
df_restructured = df.pivot_table(
    index=['country', 'year', 'age'],
    columns='sex',
    values='obs_value'
).reset_index()

# Rename columns for clarity
df_restructured.columns.name = None # Remove the 'sex' name from the columns index
df_restructured = df_restructured.rename(columns={
    'Female': 'Female_obs_value',
    'Male': 'Male_obs_value',
    'Total': 'Total_obs_value'
})

print(df_restructured.head(30
                           ))
```

```python
df['sex_age_combo'] = df['sex'] + '_' + df['age']

df_reshaped = df.pivot_table(
    index=['country', 'year'],
    columns='sex_age_combo',
    values='obs_value'
).reset_index()

df_reshaped.columns.name = None

print(df_reshaped.head())
```

```python
columns_to_keep = ['country', 'year'] + [col for col in df_reshaped.columns if '15+' in col and '15-24' not in col and '25+' not in col]
df_15_plus_only = df_reshaped[columns_to_keep]
print(df_15_plus_only.head())
```

```python
df_15_plus_only['brecha de desempleo de genero'] = df_15_plus_only['Male_15+'] - df_15_plus_only['Female_15+']
print(df_15_plus_only.head())
```
